```python
import os, random, math, warnings
import numpy as np, pandas as pd, matplotlib.pyplot as plt
import torch
from IPython.display import display, Audio
SEED=42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
DEVICE="cuda" if torch.cuda.is_available() else "cpu"
print("Device:", DEVICE)
```

```python
from transformers import pipeline, set_seed
set_seed(SEED)
gpt_generator = pipeline("text-generation", model="gpt2", device=0 if DEVICE=="cuda" else -1)

SYSTEM_PROMPT = (
"You are ShopMate, a polite E-Shop customer-support assistant. "
"You handle order tracking, returns/refunds, product information, complaints and FAQs. "
"Use simple language. Ask for missing details. Never claim to access a real order database. "
"Do not invent order status, prices, delivery dates or company policies."
)
CATEGORY_HINTS = {
"order tracking":"Ask for the order ID and explain how to check tracking.",
"returns/refunds":"Ask for order ID and reason; explain how to request a return without inventing policy.",
"product information":"Ask which product and refer to verified product specifications.",
"complaints":"Acknowledge the problem, apologize and ask for useful details.",
"FAQs":"Answer common account, payment, delivery and support questions."
}
conversation_history=[]
def shopmate_reply(user_message, category="FAQs", max_new_tokens=100):
    context="\\n".join(conversation_history[-6:])
    prompt=(SYSTEM_PROMPT+"\\nCategory: "+category+
            "\\nGuidance: "+CATEGORY_HINTS.get(category,CATEGORY_HINTS["FAQs"])+
            "\\nConversation so far:\\n"+context+
            "\\nCustomer: "+user_message+"\\nShopMate:")
    result=gpt_generator(prompt,max_new_tokens=max_new_tokens,do_sample=True,
        temperature=0.7,top_k=40,pad_token_id=50256,truncation=True)[0]["generated_text"]
    answer=result[len(prompt):].strip().split("\\nCustomer:")[0].strip()
    if not answer: answer="Could you please share a little more detail so I can help?"
    conversation_history.extend(["Customer: "+user_message,"ShopMate: "+answer])
    return answer

print(shopmate_reply("I want to know where my order is.","order tracking"))
```

```python
test_flows=[
("order tracking","My order number is ORD1024. Where can I track it?"),
("returns/refunds","The shoes I received are too small. How can I return them?"),
("product information","What details should I check before buying a laptop?"),
("complaints","My package arrived damaged. I am unhappy with this service."),
("FAQs","Which payment methods should I check on the checkout page?")
]
for category,query in test_flows:
    conversation_history.clear()
    print("\\nCATEGORY:",category,"\\nCUSTOMER:",query)
    print("SHOPMATE:",shopmate_reply(query,category))
```

```python
zero_shot="Answer as a helpful shop assistant: How do I request a return?"
few_shot=("You are a helpful E-Shop assistant.\\n"
"Customer: My parcel is late.\\nAssistant: Please check tracking in your account.\\n"
"Customer: I received the wrong item.\\nAssistant: Sorry about that. Please share your order ID.\\n"
"Customer: My product is damaged.\\nAssistant:")
role_play=("Act as ShopMate, a patient professional support representative. "
"Use simple language, acknowledge the issue and ask for needed details. "
"Never invent account information or policy. Customer: I cannot find my order. Assistant:")
for name,prompt in [("ZERO-SHOT",zero_shot),("FEW-SHOT",few_shot),("ROLE-PLAYING",role_play)]:
    result=gpt_generator(prompt,max_new_tokens=70,do_sample=True,temperature=.7,top_k=40,
                         pad_token_id=50256,truncation=True)[0]["generated_text"]
    print("\\n",name,"\\n",result[len(prompt):])
```

```python
from datasets import load_dataset
from transformers import AutoTokenizer, AutoModelForSequenceClassification, TrainingArguments, Trainer, DataCollatorWithPadding
import evaluate
MODEL_NAME="bert-base-uncased"
TRAIN_SAMPLES=2000
TEST_SAMPLES=500
imdb=load_dataset("stanfordnlp/imdb")
train_ds=imdb["train"].shuffle(seed=SEED).select(range(min(TRAIN_SAMPLES, len(imdb["train"]))))
test_ds=imdb["test"].shuffle(seed=SEED).select(range(min(TEST_SAMPLES, len(imdb["test"]))))
tokenizer=AutoTokenizer.from_pretrained(MODEL_NAME)
def tokenize_batch(batch): return tokenizer(batch["text"],truncation=True,max_length=256)
train_tok=train_ds.map(tokenize_batch,batched=True)
test_tok=test_ds.map(tokenize_batch,batched=True)
collator=DataCollatorWithPadding(tokenizer=tokenizer)
```

```python
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
from transformers import AutoModelForSequenceClassification, TrainingArguments, Trainer
import numpy as np

# 1. Initialize the sequence classification model
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=2)

# 2. Set up basic TrainingArguments
training_args = TrainingArguments(
    output_dir="./results",
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    num_train_epochs=1,
    weight_decay=0.01,
    logging_steps=50,
    fp16=torch.cuda.is_available(),
    report_to="none"
)

# 3. Define a simple metric evaluation function
def compute_metrics(eval_pred):
    metric = evaluate.load("accuracy")
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)
    return metric.compute(predictions=predictions, references=labels)

# 4. Instantiate the Trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_tok,
    eval_dataset=test_tok,
    processing_class=tokenizer,
    data_collator=collator,
    compute_metrics=compute_metrics
)

# 5. Train the model
trainer.train()

# 6. Predict and Plot results
pred = trainer.predict(test_tok)
y_true = np.array(test_tok["label"])
y_pred = np.argmax(pred.predictions, axis=1)

print(classification_report(y_true, y_pred, target_names=["negative", "positive"], digits=4))
ConfusionMatrixDisplay(confusion_matrix(y_true, y_pred), display_labels=["negative", "positive"]).plot()
plt.title("BERT – IMDB Confusion Matrix")
plt.show()
```

```python
sample_reviews = [
    "The product quality is excellent and delivery was very fast.",
    "I am disappointed because the item stopped working after one day.",
    "The packaging was good and the product matched the description.",
    "Customer service was rude and nobody solved my issue.",
    "The price is reasonable and I am happy with this purchase.",
    "The product is average, not great but not terrible.",
    "I received the wrong colour and the experience was frustrating.",
    "Amazing quality. I would definitely recommend this store.",
    "The delivery was late, but the product itself is fine.",
    "Completely useless product. I want my money back."
]

# 1. Batch BERT sentiment inference
model.to(DEVICE)
inputs = tokenizer(sample_reviews, return_tensors="pt", truncation=True, max_length=256, padding=True).to(DEVICE)
with torch.no_grad():
    logits = model(**inputs).logits
    probs_batch = torch.softmax(logits, dim=-1)

bert_labels = []
bert_confidences = []
for probs in probs_batch:
    pred_label = "positive" if int(torch.argmax(probs)) == 1 else "negative"
    confidence = float(torch.max(probs))
    bert_labels.append(pred_label)
    bert_confidences.append(round(confidence, 3))

# 2. Batch GPT chatbot response generation
# Construct batch prompts matching the previous logic
prompts = []
for review in sample_reviews:
    prompt = (SYSTEM_PROMPT + "\nCategory: complaints" +
              "\nGuidance: " + CATEGORY_HINTS["complaints"] +
              "\nConversation so far:\n" +
              "\nCustomer: " + review + "\nShopMate:")
    prompts.append(prompt)

# Generate responses in batch
generated_outputs = gpt_generator(
    prompts,
    max_new_tokens=65,
    do_sample=True,
    temperature=0.7,
    top_k=40,
    pad_token_id=50256,
    truncation=True
)

# Extract and clean generated answers
gpt_replies = []
for prompt, output in zip(prompts, generated_outputs):
    result = output[0]["generated_text"]
    answer = result[len(prompt):].strip().split("\nCustomer:")[0].strip()
    if not answer:
        answer = "Could you please share a little more detail so I can help?"
    gpt_replies.append(answer)

# 3. Compile and display comparison DataFrame
rows = []
for idx, review in enumerate(sample_reviews):
    rows.append([
        idx + 1,
        review,
        gpt_replies[idx],
        bert_labels[idx],
        bert_confidences[idx]
    ])

comparison_df = pd.DataFrame(rows, columns=["No.", "Customer review", "GPT chatbot response", "BERT sentiment", "BERT confidence"])
display(comparison_df)
```
